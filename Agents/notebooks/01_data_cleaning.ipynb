```python
import pandas as pd

# Загрузка
df = pd.read_csv("crypto_news_sentiment.csv")

# Объединяем title и text в один текст
df["text"] = df["title_news"].fillna('') + ". " + df["text_news"].fillna('')

# Чистим (удаляем URL, эмодзи, хэштеги)
import re
def clean_text(text):
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"#\S+", "", text)
    text = re.sub(r"@\S+", "", text)
    text = re.sub(r"[^A-Za-z0-9.,!?$%&\s]", "", text)
    return text.strip()

df["text"] = df["text"].apply(clean_text)
df = df[df["text"].str.len() > 20]  # убираем пустые строки
```
